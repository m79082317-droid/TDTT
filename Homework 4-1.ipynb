W4A25
Nhập vào một dãy số nguyên, ngưng nhập khi người dùng nhập -1.
Sau khi nhập xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập

```python
max_val = None
min_val = None
while True:
    n = int(input("Nhập số: "))
    if n == -1:
        break
    if max_val is None:
        max_val = n
        min_val = n
    else:
        if n > max_val:
            max_val = n 
            if n < min_val:
            min_val = n
if max_val is not None:
    print("--- Kết quả ---")
    print(f"Số lớn nhất: {max_val}")
    print(f"Số nhỏ nhất: {min_val}")
```

W4A26
[Fibonaci1]
Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ
bằng tổng hai số trước đó
Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt
quá A

```python
def find_fibonacci(A):
    if A < 1:
        return None
    f1, f2 = 1, 1
    current_fib = 1
    while f2 <= A:
        current_fib = f2
        # Tính số tiếp theo trong dãy
        f1, f2 = f2, f1 + f2
    return current_fib
try:
    A = int(input("Nhập số nguyên dương A: "))
    result = find_fibonacci(A)
    print(f"Số Fibonacci lớn nhất không vượt quá {A} là: {result}")
except ValueError:
    print("Vui lòng nhập một số nguyên dương.")
```

W4A27
Nhập vào một chuỗi, hãy đếm xem trong chuỗi có bao nhiêu từ (quy định là chuỗi không có
ký tự đặc biệt, không số, không

```python
def dem_so_tu_bang_vong_lap(chuoi):
    so_tu = 0
    dang_trong_tu = False
    for ky_tu in chuoi:
        if ky_tu != " " and not dang_trong_tu:
            so_tu += 1
            dang_trong_tu = True 
        elif ky_tu == " ":
            dang_trong_tu = False
            
    return so_tu

# Nhập và kiểm tra
van_ban = input("Nhập vào chuỗi: ")
print(f"Số từ đếm được là: {dem_so_tu_bang_vong_lap(van_ban)}")
```

Nhập vào một chuỗi, hãy in từ đầu tiên trong chuỗi
Dùng cách: for trên index; KHÔNG dùng hàm split()

```python
s = input("Nhập vào chuỗi: ")
for i in range(len(s)):
    if s[i] != " ":
        start_index = i
        break
if start_index != -1:
    end_index = len(s) 
    for j in range(start_index, len(s)):
        if s[j] == " ":
            end_index = j
            break
    tu_dau_tien = s[start_index:end_index]
    print(f"Từ đầu tiên là: {tu_dau_tien}")
else:
    print("Chuỗi không có từ nào.")
```

W4A29
Cho một chuỗi chỉ gồm các ký tự chữ cái thường và chữ cái hoa, hãy đổi ký tự hoa thành
thường và thường thành hoa.

```python
def swap_case_manual(s):
    res = ""
    for char in s:
        if char.isupper():
            res += char.lower()
        elif char.islower():
            res += char.upper()
        else:
            res += char
    return res
```
