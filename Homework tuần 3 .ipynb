Bài 12: Nhập số nguyên dương n, Cho S(k) = 1 + 2 + 3 + … + k. Tìm k sao cho S(k)
lớn nhất nhưng nhỏ hơn n

```python
n = int(input("Nhap so nguyen duong n: "))
S = 0
k = 0
while S + (k + 1) < n:
    k += 1
    S += k
print(f"Gia tri k lon nhat la: {k}")
print(f"Tong S({k}) la: {S} (nho hon {n})")
```

Bài 13: Nhập vào số nguyên dương A, tìm n nhỏ nhất sao cho
1 + 1/2 + 1/3 + 1/4 + ... + 1/n > A 

```python
A = int(input("Nhap vao so nguyen duong A: "))
S = 0
n = 0
while S <= A:
    n += 1
    S += 1 / n
print(f"So n nho nhat la: {n}")
print(f"Tong thuc te S = {S:.4f} (lon hon {A})")
```

Bài 14: Nhập vào một dãy số nguyên, dừng nhập khi người dùng nhập -1. Sau khi nhập
xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập. KHÔNG ĐƯỢC dùng mảng

```python
max = None
min = None
print("Nhap day so(Nh) :")
while True:
    try:
        n = int(input("Nhap so: "))
        if n == -1:
            break
        if max is None:
            max = n
            min = n
        else:
            if n > max:
                max = n
            if n < min:
                min = n
    except ValueError:
        print("Chi nhap so nguyen!")
if max is not None:
    print(f"Max: {max}")
    print(f"Min: {min}")
else:
    print("Chưa nhap.")
```

Bài 15: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số. Làm bằng nhiều
cách nhất có thể, ko dùng mảng.

```python
n = int(input("Nhap n: "))
temp = abs(n)
count = 0

if temp == 0:
    count = 1
else:
    while temp > 0:
        temp //= 10
        count += 1
print(f"So chu so: {count}")
```

Bài 19: Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ bằng tổng hai số trước đó. Nhập vào số nguyên dương A, hãy tìm số trongdãy số fibonacci lớn nhất nhưng không vượt quá A

```python
A= int(input("Nhap so A:"))
a, b = 1, 1
if a < 1:
    print("Khong co so Fibonacci nao")
else:
    while a + b<= a:
        c=a+b
        a=b
        b=c
print(f"So Fibonacci lon nhat khong vuot qua {A} la: {b}")        
```
