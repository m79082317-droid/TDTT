Cho hai số nguyên dương, n và m (0 < n, m <= 20 ). .Bạn hãy in ra tất cả các chuỗi nhị phân có
n số 0 và m số 1. In theo thứ tự tăng dần trong từ điển

```python
import sys
sys.setrecursionlimit(2000)
def generate_binary_strings(n, m, current_string=""):
    if n == 0 and m == 0:
        print(current_string)
        return
    if n > 0:
        generate_binary_strings(n - 1, m, current_string + '0')
    if m > 0:
        generate_binary_strings(n, m - 1, current_string + '1')
def main():
    try:
        line = sys.stdin.read().split()
        if len(line) >= 2:
            n = int(line[0])
            m = int(line[1])
            generate_binary_strings(n, m)
    except Exception as e:
        pass
if __name__ == "__main__":
    main()
```

W9A17 Viết hàm tìm dãy con tăng nghiêm ngặt dài nhất trong dãy đầu vào từ bàn phím

```python
def binary_search_left(tails, target):
    low = 0
    high = len(tails) - 1
    result = len(tails)
    while low <= high:
        mid = (low + high) // 2
        if tails[mid] >= target:
            result = mid      
            high = mid - 1    
        else:
            low = mid + 1     
    return result
def find_lis_no_import(arr):
    if not arr:
        return 0, []
    tails = []         
    tail_indices = []   
    parent = [-1] * len(arr) 
    for i, x in enumerate(arr):
        idx = binary_search_left(tails, x)
        if idx == len(tails):
            tails.append(x)
            tail_indices.append(i)
        else:
            tails[idx] = x
            tail_indices[idx] = i
        if idx > 0:
            parent[i] = tail_indices[idx - 1]
    lis_length = len(tails)
    lis_subseq = []
    if lis_length > 0:
        curr = tail_indices[-1]
        while curr != -1:
            lis_subseq.append(arr[curr])
            curr = parent[curr]
        lis_subseq = lis_subseq[::-1]
    return lis_length, lis_subseq
if __name__ == "__main__":
    try:
        input_str = input("Nhập các phần tử của dãy (cách nhau bởi dấu cách): ")
        arr = [int(x) for x in input_str.split()]
        if not arr:
            print("Dãy rỗng!")
        else:
            length, subseq = find_lis_no_import(arr)
            print("\n--- Kết quả ---")
            print(f"Dãy ban đầu: {arr}")
            print(f"Độ dài dãy con tăng nghiêm ngặt dài nhất: {length}")
            print(f"Một dãy con hợp lệ là: {subseq}")
    except ValueError:
        print("Vui lòng chỉ nhập các số nguyên hợp lệ!")
```

Thực hiện chuẩn hóa chuỗi kí tự:
a. bỏ khoảng trắng đầu cuối
b. các khoảng trắng giữa chỉ còn 1 dấu cách
c. viết hoa chữ cái đầu mỗi từ
KHÔNG DÙNG CÁC HÀM SẴN CỐ

```python
def chuan_hoa_chuoi(s):
    ket_qua = []
    chu_dau = True 
    for i in range(len(s)):
        char = s[i]
        if char != ' ' and char != '\t' and char != '\n':
            ma_ascii = ord(char)
            if chu_dau:
                if 97 <= ma_ascii <= 122:
                    char = chr(ma_ascii - 32)
                chu_dau = False
            else:
                if 65 <= ma_ascii <= 90:
                    char = chr(ma_ascii + 32)
            ket_qua.append(char)
        else:
            if len(ket_qua) > 0 and ket_qua[-1] != ' ':
                ket_qua.append(' ')
                chu_dau = True 
    if len(ket_qua) > 0 and ket_qua[-1] == ' ':
        ket_qua.pop() 
    chuoi_chuan_hoa = ""
    for c in ket_qua:
        chuoi_chuan_hoa += c
    return chuoi_chuan_hoa
```

W9A21
Tìm phần tử lớn thứ k trong dãy n phần tử bằng thuật toán tốt nhất có thể – Tránh dùng sắp xếp
(dùng phân hoạch của quicksort)

```python
def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1
    for j in range(low, high):
        if arr[j] > pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1
def find_kth_largest(arr, k):
    low = 0
    high = len(arr) - 1
    target_idx = k - 1 
    while low <= high:
        pivot_idx = partition(arr, low, high)
        if pivot_idx == target_idx:
            return arr[pivot_idx]
        elif pivot_idx > target_idx:
            high = pivot_idx - 1
        else:
            low = pivot_idx + 1
    return None
```

W9A22
Hai chuỗi là anagram nếu chúng có cùng ký tự và số lần xuất hiện. viết CT kiểm tra xem 2 chuỗi
đầu vào có phải là anagram của nhau không

```python
def is_anagram(str1: str, str2: str) -> bool:
    s1 = str1.replace(" ", "").lower()
    s2 = str2.replace(" ", "").lower()
    if len(s1) != len(s2):
        return False
    char_count = {}
    for char in s1:
        if char in char_count:
            char_count[char] += 1
        else:
            char_count[char] = 1
    for char in s2:
        if char in char_count:
            char_count[char] -= 1
        else:
            return False
    for count in char_count.values():
        if count != 0:
            return False
    return True
```

W9A23
Cho số nguyên rất lớn (string), và một số nguyên dương k. Tìm số có k chữ số lớn nhất có thể
từ số ban đầu bằng cách xóa bớt đi một số chữ số nào đó mà không thay đổi vị trí của bất kì
chữ số nào. (VD: 78934340999, k = 7 => 9440999 là số lớn nhất có 7 chữ số cần tìm)

```python
def find_largest_number(s: str, k: int) -> str:
    rem = len(s) - k  
    stack = []
    for ch in s:
        while rem > 0 and stack and stack[-1] < ch:
            stack.pop()
            rem -= 1
        stack.append(ch)
    final_result = stack[:k]
    return "".join(final_result)
```

W9A24
Nén chuối kí tự bằng cách đếm số lần xuất hiện của các kí tự liên tiếp nhau và ghi lại ngay sau
kí tự đó. VD aaaaaafdhhhhh =>a6f1d1h5

```python
def nen_chuoi(s: str) -> str:
    if not s:
        return ""
    ket_qua = []
    dem = 1
    for i in range(len(s) - 1):
        if s[i] == s[i + 1]:
            dem += 1
        else:
            ket_qua.append(f"{s[i]}{dem}")
            dem = 1 
    ket_qua.append(f"{s[-1]}{dem}")
    return "".join(ket_qua)
```

Giả sử mật khẩu mạnh là một chuỗi kí tự dài ít nhất 8 kí tự gồm ít nhất 1 chữ cái thường, 1 chữ
cái hoa, 1 chữ số và 1 trong các kí tự đặc biệt !, @, #, $. Nhập vào một chuỗi kĩ tự, kiểm tra xem
nó có phải là mật khẩu mạnh không

```python
def kiem_tra_mat_khau(password):
    if len(password) < 8:
        return False
    co_chu_thuong = False
    co_chu_hoa = False
    co_chu_so = False
    co_ky_tu_dac_biet = False
    ky_tu_cho_phep = "!@#$"
    for ky_tu in password:
        if ky_tu.islower():     
            co_chu_thuong = True
        elif ky_tu.isupper():     
            co_chu_hoa = True
        elif ky_tu.isdigit():    
            co_chu_so = True
        elif ky_tu in ky_tu_cho_phep: 
            co_ky_tu_dac_biet = True
    if co_chu_thuong and co_chu_hoa and co_chu_so and co_ky_tu_dac_biet:
        return True
    else:
        return False
chuoi_nhap = input("Nhập vào chuỗi mật khẩu cần kiểm tra: ")

if kiem_tra_mat_khau(chuoi_nhap):
    print("MẠNH!")
else:
    print("YẾU")
```
