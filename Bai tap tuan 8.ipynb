Cho một mảng số nguyên không âm. Gọi
mảng đó là "mảng đẹp" nếu mỗi phần tử
x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp
hay không. Nếu không hãy xóa đi y số (y
là nhỏ nhất có thể) để mảng còn lại thành
mảng đẹp

```python
def day_so_dep(arr):
    counts = {}
    for x in arr:
        if x in counts:
            counts[x] += 1
        else:
            counts[x] = 1 
    y = 0
    for x, c in counts.items():
        if c >= x:
            y += (c - x)
        else:
            y += c
    if y == 0:
        print("Mảng đã là mảng đẹp!")
    else:
        print(f"Cần xóa ít nhất {y} số để mảng thành mảng đẹp.")
    return y
```

Merge 2 mảng đã sắp xếp (có thể tăng
dần, có thể giảm dần) mà không dùng
bất cứ thuật toán sắp xếp nào thành một
mảng cũng có cùng thứ tự sắp xếp như 2
mảng ban đầu


```python
arr1 = [1, 3, 5, 7]
arr2 = [2, 4, 6, 8, 9]
len1 = 0
for element in arr1:
    len1 += 1
len2 = 0
for element in arr2:
    len2 += 1
is_ascending = True
if len1 > 1:
    if arr1[0] > arr1[len1 - 1]:
        is_ascending = False
elif len2 > 1:
    if arr2[0] > arr2[len2 - 1]:
        is_ascending = False
total_len = len1 + len2
result = [0] * total_len
i = 0  
j = 0  
k = 0  
while i < len1 and j < len2:
    if is_ascending:
        if arr1[i] <= arr2[j]:
            result[k] = arr1[i]
            i += 1
        else:
            result[k] = arr2[j]
            j += 1
    else:
        if arr1[i] >= arr2[j]:
            result[k] = arr1[i]
            i += 1
        else:
            result[k] = arr2[j]
            j += 1
    k += 1
while i < len1:
    result[k] = arr1[i]
    i += 1
    k += 1
while j < len2:
    result[k] = arr2[j]
    j += 1
    k += 1
print(result)
```

Cho N tên sinh viên gồm cả họ và tên,
tìm xem tên nào hay gặp nhất. Nếu có
nhiều tên cùng số lần gặp thì lấy tên có
thứ tự từ điển nhỏ nhất

```python
n = int(input())
name_counts = {}
for _ in range(n):
    name = input()
    if name in name_counts:
        name_counts[name] += 1
    else:
        name_counts[name] = 1
best_name = ""
max_count = 0
for name in name_counts:
    count = name_counts[name]
    if count > max_count:
        max_count = count
        best_name = name
    elif count == max_count:
        if name < best_name:
            best_name = name
print(best_name)
```

```python
num_str = input()
k = int(input())
n = 0
for _ in num_str:  
    n += 1
to_remove = n - k  
stack = [""] * n
stack_size = 0  
for i in range(n):
    digit = num_str[i]
    while to_remove > 0 and stack_size > 0 and stack[stack_size - 1] < digit:
        stack_size -= 1  
        to_remove -= 1   
    stack[stack_size] = digit
    stack_size += 1
result = ""
for i in range(k):
    result += stack[i]
print("Số lớn nhất có", k, "chữ số là:", result)
```

Bài toán Tìm dãy các hợp số dài nhất liên
tiếp không chứa số nguyên nào <= N (N
là số nguyên dương cho trước). Nếu có
nhiều dãy cùng độ dài thì lấy dãy nhỏ
nhất (gặp đầu tiên từ trái qua)

```python
N = int(input("Nhập số N: "))
LIMIT = 100000
is_prime = [True] * (LIMIT + 1)
is_prime[0] = is_prime[1] = False
for i in range(2, int(LIMIT**0.5) + 1):
    if is_prime[i]:
        for j in range(i * i, LIMIT + 1, i):
            is_prime[j] = False
max_len = 0          
current_len = 0      
start_num = -1       
best_start_num = -1  
for i in range(N + 1, LIMIT + 1):
    if not is_prime[i]:
        if current_len == 0:
            start_num = i  
        current_len += 1
    else:
        if current_len > max_len:
            max_len = current_len
            best_start_num = start_num
        current_len = 0  
if current_len > max_len:
    max_len = current_len
    best_start_num = start_num
print("--- KẾT QUẢ ---")
if max_len > 0:
    print(f"Dãy hợp số dài nhất bắt đầu từ số: {best_start_num}")
    print(f"Độ dài của dãy: {max_len}")
    print("Các số trong dãy là: ", end="")
    for k in range(best_start_num, best_start_num + max_len):
        print(k, end=" ")
    print()
```

Tìm Dãy con tăng dần liên tục dài nhất
của một danh sách nhập vào từ bàn
phím. Lấy dãy đầu tiên nếu có nhiều dãy
có số phần tử bằng nhau

```python
user_input = input().split()
arr = [int(x) for x in user_input.split()]
a = 0      
b = 1        
current_start = 0   
current_len = 1     
for i in range(1, len(arr)):
    if arr[i] > arr[i - 1]:
        current_len += 1
    else:
        if current_len > b:
            a = current_len
            a = current_start
        current_start = i
        current_len = 1
if current_len > b:
    b = current_len
    a = current_start
dai_nhat = arr[a : a + b]
print(dai_nhat)
```

W7A30
Tìm Dãy con tăng dần KHÔNG liên tục
dài nhất của một danh sách nhập vào từ
bàn phím

```python
input_str = input()
arr = [int(x) for x in input_str.split()]
n = len(arr)
L = [1] * n
trace = [-1] * n
for i in range(1, n):
    for j in range(0, i):
        if arr[i] > arr[j] and L[i] < L[j] + 1:
            L[i] = L[j] + 1
            trace[i] = j
        max_length = max(L)
        max_index = L.index(max_length)
        lis = []
        while max_index != -1:
            lis.append(arr[max_index])
            max_index = trace[max_index]
        lis.reverse()
print(f"Độ dài dãy con tăng dài nhất: {max_length}")
print(f"Dãy con đó là: {lis}")
```

```python
n_str = input("Nhập số cần đổi chỗ: ")
s = list(n_str)
length = len(s)
da_doi_cho = False
for i in range(length):
    min_digit = s[i]
    min_idx = i
    for j in range(length - 1, i, -1):
        if i == 0 and s[j] == '0':
            continue
        if s[j] < min_digit:
            min_digit = s[j]
            min_idx = j
    if min_idx != i:
        s[i], s[min_idx] = s[min_idx], s[i]
        da_doi_cho = True
        break
ket_qua = int("".join(s))
if da_doi_cho:
    print(ket_qua)
else:
    print(ket_qua)
```
