W4A7"[LargestPrimeFactor]
Nhập vào số nguyên dương n (n ≥ 2). Hãy in ra ước số nguyên tố lớn nhất của n."

```python
n = int(input("Nhập số nguyên dương n: "))
steps = 0
while True:
    temp = n
    reversed_n = 0
    while temp > 0:
        remainder = temp % 10
        reversed_n = (reversed_n * 10) + remainder
        temp = temp // 10
    if n == reversed_n:
        break
    n = n + reversed_n
    steps += 1
print(f"Số bước: {steps}")
print(f"Số đối xứng thu được: {n}")
```

W4A8[ReverseAddPalindrome]
Nhập số nguyên dương n. Lặp lại phép toán n = n + reverse(n) cho đến khi n là số đối xứng (palindrome). In ra số bước và giá trị palindrome thu được.

```python
n = int(input("Nhập n: "))
buoc = 0
while str(n) != str(n)[::-1]: 
    dao_nguoc = int(str(n)[::-1]) 
    n = n + dao_nguoc             
    buoc += 1                     
    print(f"Bước {buoc}: n = {n}")
print("--- Kết quả cuối cùng ---")
print(f"Số đối xứng: {n}")
print(f"Tổng số bước: {buoc}")
```

W4A10
"[LongestCollatzUpToN]
Nhập vào số nguyên dương n. Tìm số x trong [1, n] có độ dài dãy Collatz lớn nhất và in ra x cùng độ dài dãy tương ứng (x L). Nếu có nhiều số cùng độ dài, chọn số nhỏ nhất."

```python
line = sys.stdin.readline()
if line:
    n = int(line.strip())
    max_len = 0
    best_x = 0
    memo = {1: 1}
    for i in range(1, n + 1):
        curr = i
        temp_path = []
        while curr not in memo:
            temp_path.append(curr)
            if curr % 2 == 0:
                curr //= 2
            else:
                curr = 3 * curr + 1
        current_length = memo[curr]
        for j in range(len(temp_path) - 1, -1, -1):
            current_length += 1
            memo[temp_path[j]] = current_length
        if memo[i] > max_len:
            max_len = memo[i]
            best_x = i
    print(f"{best_x} {max_len}")
```
