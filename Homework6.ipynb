W6A21
Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ
chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện

```python
def map_string_positions(s):
    a = s.split()
    pos_dict = {}
    for index, word in enumerate(a):
        if word not in pos_dict:
            pos_dict[word] = []
        pos_dict[word].append(index)
    return {k: tuple(v) for k, v in pos_dict.items()}
input_str = input("Nhập chuỗi: ")
result = map_string_positions(input_str)
print(result)
```

W6A28
Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm
thực sự, hay vô danh (không tăng, không giảm)

```python
def Day_tang_giam(a)
    if len(a) < 2:
        return "Dãy quá ngắn để kết luận"
    is_increasing = True
    is_decreasing = True
    for i in range(len(a) - 1):
        if a[i] >= a[i+1]:
            is_increasing = False
        if a[i] <= a[i+1]:
            is_decreasing = False
    if is_increasing:
        return "Dãy tăng thực sự"
    elif is_decreasing:
        return "Dãy giảm thực sự"
    else:
        return "Dãy vô danh"
```
