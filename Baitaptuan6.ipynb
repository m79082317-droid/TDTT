W6A4
Nhập một dòng gồm các cặp key:value cách nhau bởi dấu cách. Tạo
dictionary ánh xạ mỗi key -> list các value theo thứ tự xuất hiện. Nếu key
lặp lại, thêm vào list của key đó

```python
input_data = input().split()
result_dict = {}
for item in input_data:
    key, value = item.split(':', 1)
    if key in result_dict:
        result_dict[key].append(value)
    else:
        result_dict[key] = [value]
print(result_dict)
```

W6A5
Nhập một list số nguyên, In ra dictionary {'positives': ..., 'negatives': ...,
'zeros': ...} (đếm số >0, <0, =0)

```python
numbers = list(map(int, input().split()))
counts = {'positives': 0,'negatives': 0,'zeros': 0}
for num in numbers:
    if num > 0:
        counts['positives'] += 1
    elif num < 0:
        counts['negatives'] += 1
    else:
        counts['zeros'] += 1
print(counts)
```

W6A7
Viết chương trình nhập vào một tuple, in ra phần tử đầu, phần tử cuối, và
tuple đảo ngược

```python
input_data = input("Nhập các phần tử của tuple (cách nhau bởi dấu cách): ")
my_tuple = tuple(input_data.split())
if len(my_tuple) > 0:
    first_element = my_tuple[0]
    last_element = my_tuple[-1]
    reversed_tuple = my_tuple[::-1]
    print(f"Phần tử đầu: {first_element}")
    print(f"Phần tử cuối: {last_element}")
    print(f"Tuple đảo ngược: {reversed_tuple}")
else:
    print("Tuple trống!")
```
