kiểu số nguyên, đơn vị VND).
Các quy tắc cần cài đặt:
1. Giá mở cửa (0–0.8 km):
- Xe 4 chỗ: 11,000 VND
- Xe 7 chỗ: 12,000 VND
2. Giá theo km sau 0.8 km:
- Xe 4 chỗ: từ 0.8→20 km ơnh 13,000 VND/km; trên 20 km tính 10,500 VND/km
- Xe 7 chỗ: từ 0.8→20 km ơnh 14,000 VND/km; trên 20 km tính 12,000 VND/km
3. Phí chờ: 5 phút đầu miễn phí; sau đó, mỗi 5 phút (làm tròn lên theo từng block 5 phút) tính 2,000 VND.
4. Phụ thu ban đêm: nếu start_hour thuộc [22,23,0,1,2,3,4,5], cộng +20% trên tiền tạm tính (mở cửa + quãng đường +
chờ).
5. Điều kiện đặc biệt:
- Nếu km > 30 và không phải ban đêm, áp dụng giảm 10% trên tổng tiền.
- Nếu tổng tiền cuối cùng < 50,000 nhưng km > 5, đặt tổng tiền bằng 50,000 VND (mức tối thiểu). 

```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    if car_type==4:
        a=11000
    elif car_type==7:
        a=12000
    if 0.8<=km<=20:
        if car_type==4:
            b=km*13000
        if car_type==7:
            b=km*14000
    if km>20:
        if car_type==4:
            b=260000+(km-20)*10500
        if car_type==7:
            b=280000+(km-20)*12000
    if wait_minutes<=5:
        c=0
    else:
        m=wait_minutes%5
        if m!=0:
            c=(wait_minutes//5)*2000+2000-2000
        else:
            c=(wait_minutes/5)*2000-2000
    Tong=a+b+c
    if start_hour in [22,23,0,1,2,3,4,5]:
        Tong=Tong*1.2
    else:
        if km>30:
            Tong=Tong*0.9
    if Tong<50000 and km>5:
        Tong=50000
    return Tong
car_type, start_hour, km, wait_minutes=map(int,input().split())
print(smart_taxi_fare(car_type, start_hour, km, wait_minutes))
```

16 Viết hàm trả về góc ∠ABC theo đơn vị độ khi biết tọa độ các điểm A(x1,y1), B(x2,y2), C(x3,y3).

```python
def tinh_goc_abc(x1, y1, x2, y2, x3, y3):
    def my_sqrt(n):
        if n == 0: return 0
        x = n
        y = (x + 1) / 2
        while y < x:
            x = y
            y = (x + n / x) / 2
        return x
    a2 = (x3 - x2)**2 + (y3 - y2)**2
    c2 = (x1 - x2)**2 + (y1 - y2)**2
    b2 = (x3 - x1)**2 + (y3 - y1)**2
    if a2 == 0 or c2 == 0:
        return 0.0
    a = my_sqrt(a2)
    c = my_sqrt(c2)
    cos_B = (a2 + c2 - b2) / (2 * a * c)
    if cos_B > 1.0: cos_B = 1.0
    if cos_B < -1.0: cos_B = -1.0
    PI = 3.141592653589793
    def my_acos(x):
        res = 0.0
        if x > 0.7:
            res = my_sqrt(2 * (1 - x)) * (1 + (1 - x) / 12)
            return res
        elif x < -0.7:
            res = PI - my_sqrt(2 * (1 + x)) * (1 + (1 + x) / 12)
            return res
        else: 
            return PI / 2 - (x + x**3 / 6 + 3 * x**5 / 40)
    angle_rad = my_acos(cos_B)
    angle_deg = angle_rad * 180 / PI
    return angle_deg
```

- Đầu vào:
- quad: danh sách hoặc bộ gồm đúng 4 đỉnh [(x1,y1), (x2,y2), (x3,y3), (x4,y4)] theo thứ tự (tứ giác có thể lồi hoặc lõm,

- p: một điểm (x, y).
- Đầu ra: giá trị boolean.
- Gợi ý: Cài đặt phép kiểm tra điểm trong đa giác tổng quát (ví dụ: ray casting), đồng thời tính cả trường hợp điểm nằm
trên biên là ở trong.

```python
def point_in_quadrilateral(quad, p):
    x, y = p
    n = len(quad)
    inside = False
    for i in range(n):
        p1 = quad[i]
        p2 = quad[(i + 1) % n]
        if (min(p1[1], p2[1]) <= y <= max(p1[1], p2[1]) and
            min(p1[0], p2[0]) <= x <= max(p1[0], p2[0])):
            cross_product = (p2[1] - p1[1]) * (x - p1[0]) - (p2[0] - p1[0]) * (y - p1[1])
            if abs(cross_product) < 1e-9: 
                return True
        if ((p1[1] > y) != (p2[1] > y)):
            intersect_x = (p2[0] - p1[0]) * (y - p1[1]) / (p2[1] - p1[1]) + p1[0]
            if x < intersect_x:
                inside = not inside
    return inside
hinh_chu_nhat = [(0, 0), (4, 0), (4, 3), (0, 3)]
diem_trong = (2, 2)
diem_tren_bien = (4, 1.5)
diem_ngoai = (5, 5)
print(f"Điểm {diem_trong} trong tứ giác: {point_in_quadrilateral(hinh_chu_nhat, diem_trong)}")
print(f"Điểm {diem_tren_bien} trên biên: {point_in_quadrilateral(hinh_chu_nhat, diem_tren_bien)}")
print(f"Điểm {diem_ngoai} ngoài tứ giác: {point_in_quadrilateral(hinh_chu_nhat, diem_ngoai)}")
```

```python
def are_collinear(x1, y1, x2, y2, x3, y3):
    A = x1 * (y2 - y3) + x2 * (y3 - y1) + x3 * (y1 - y2)
    B = 1e-9
    return abs(A) < B
```

chấm động, ngược lại trả về 'neither'.
- Đầu vào: chuỗi s.
- Đầu ra: một trong các giá trị {'int', 'float', 'neither'}.
- Quy tắc:
- Cho phép có dấu + hoặc - ở đầu.
- Số nguyên phải chỉ gồm các chữ số và không có dấu chấm thập phân (ví dụ: '100').
- Số thực phải có dấu chấm và có chữ số ở cả hai phía của dấu chấm (ví dụ: '100.0', '-3.14').
- Các chuỗi như '.5', '5.', '1e3' được xem là 'neither' trong bài này.

```python
def detect_number_type(s):
    temp_s = s
    if temp_s.startswith('+') or temp_s.startswith('-'):
        temp_s = temp_s[1:]
    if not temp_s:
        return 'neither'
    if temp_s.isdigit():
        return 'int'
    
    # Kiểm tra số thực: 
    # 1. Phải chứa đúng một dấu chấm
    # 2. Hai phần phân tách bởi dấu chấm phải toàn chữ số và không được rỗng
    if '.' in temp_s:
        parts = temp_s.split('.')
        if len(parts) == 2 and parts[0].isdigit() and parts[1].isdigit():
            return 'float'
            
    return 'neither'

# --- Kiểm tra nhanh ---
test_cases = ['100', '-100', '3.14', '-3.14', '.5', '5.', '1e3', 'abc', '+100.0']
for t in test_cases:
    print(f"'{t}': {detect_number_type(t)}")
```

31 Yêu cầu: Viết hàm is_perfect_square(n) trả về True nếu n là số chính phương, ngược lại trả về False.
- Đầu vào: số nguyên n (có thể âm).
- Đầu ra: giá trị boolean.
- Lưu ý: Số âm không phải là số chính phương trong tập số nguyên.

```python
def is_perfect_square(n):
    if n < 0:
        return False
    if n < 2:
        return True
    root = int(n ** 0.5)
    A= root * root == n
    return A
```

32 Yêu cầu: Viết hàm is_right_triangle(a, b, c) trả về True nếu a, b, c có thể là ba cạnh của một tam giác vuông, ngược lại trả
về False.
- Đầu vào: các số dương a, b, c.
- Đầu ra: giá trị boolean.
- Lưu ý: Kiểm tra bất đẳng thức tam giác và định lý Pythagore. Dùng sai số nhỏ khi so sánh số thực.

```python
def is_right_triangle(a, b, c):
    sides = sorted([a, b, c])
    s1, s2, s3 = sides[0], sides[1], sides[2]
    if s1 + s2 <= s3:
        return False
    eps = 1e-9
    return abs(s1**2 + s2**2 - s3**2) < eps
```

33 Yêu cầu: Viết hàm compute(a, b, op) tính a (+|-||/) b tùy theo op.
- Đầu vào: các số a, b và ký tự op thuộc {'+', '-', '', '/'}.
- Đầu ra: kết quả số học.
- Yêu cầu thêm:
- Nếu op là '/' và b == 0, phát sinh ValueError.
- Nếu op không hợp lệ, phát sinh ValueError.

```python
def compute(a, b, op):
    if op == '+':
        return a + b
    elif op == '-':
        return a - b
    elif op == '*':
        return a * b
    elif op == '/':
        if b == 0:
            return False
        return a / b
```

35 Yêu cầu: Viết hàm divide_with_remainder(dividend, divisor) trả về bộ (q, r) trong đó q là thương nguyên và r là số dư sao
cho dividend = divisorq + r với 0 <= |r| < |divisor| (dùng ngữ nghĩa của Python với // và %).
- Đầu vào: số nguyên dividend, divisor (divisor khác 0).
- Đầu ra: bộ hai số nguyên (q, r).
- Ví dụ: divide_with_remainder(5, 3) -> (1, 2).

```python
def divide_with_remainder(dividend, divisor):
    q = dividend // divisor
    r = dividend % divisor
    return (q, r)
```

41 Viết hàm trả về ký tự đứng sau ký tự x n vị trí trong bảng chữ cái.

```python
def shift_character(x, n):
    if 'A' <= x <= 'Z':
        return chr((ord(x) - ord('A') + n) % 26 + ord('A'))
    elif 'a' <= x <= 'z':
        return chr((ord(x) - ord('a') + n) % 26 + ord('a'))
    else:
        return x
```

43 Viết hàm tính diện tích tam giác có ba cạnh a, b, c (dùng công thức Heron). Phát sinh ValueError nếu ba cạnh không thể
tạo thành tam giác.

```python
def tinh_dien_tich_tam_giac(a, b, c):
    if a + b <= c or a + c <= b or b + c <= a:
        raise ValueError(f"Ba cạnh ({a}, {b}, {c}) không thỏa mãn bất đẳng thức tam giác.")  
    if a <= 0 or b <= 0 or c <= 0:
        raise ValueError("Độ dài các cạnh phải là số dương.")
    p = (a + b + c) / 2
    dien_tich = (p * (p - a) * (p - b) * (p - c)) ** 0.5   
    return dien_tich
```

45 Viết hàm trả về số lượng số nguyên dương k sao cho 1 <= k < n và k > (n - 1) / 2. (Ví dụ: n=9 -> 4).

```python
def count_k(n):
    if n <= 1:
        return 0
    lower_bound = (n - 1) / 2
    import math
    start_k = math.floor(lower_bound) + 1
    end_k = n - 1
    if start_k > end_k:
        return 0
    return end_k - start_k + 1
print(f"Kết quả với n=9: {count_k(9)}") # Output: 4
```
