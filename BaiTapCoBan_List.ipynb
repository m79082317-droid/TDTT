### Bài 7: Kiểm tra một list số nguyên có phải là dãy tăng, dãy giảm hay vô danh
1 3 5 6 7 8 9: Tang
9 8 7 6 5 4 0: Giam
1 2 3 7 1 0 2: Vo danh
1 2 2 4 5 7: Vo danh

```python
def tang_giam(arr):
    if len(arr) < 2:
        return "Tang" 
    is_increasing = True
    is_decreasing = True

    for i in range(len(arr) - 1):
        if arr[i] >= arr[i+1]:
            is_increasing = False
        if arr[i] <= arr[i+1]:
            is_decreasing = False
    if is_increasing:
        return "Tang"
    elif is_decreasing:
        return "Giam"
    else:
        return "Vo danh"
lst=list(input())
print(tang_giam(lst))
```

### Bài 8: Kiểm tra một list số nguyên có phải là ziczac khong
1 3 2 5 2 8 7 9 2 100: ziczac
34 2 56 7 90 20 30 7 120: ziczac

```python
def is_ziczac(arr):
    n = len(arr)
    if n < 2:
        return True    
    for i in range(1, n - 1):
        prev = arr[i-1]
        curr = arr[i]
        nxt = arr[i+1]
        if not ((prev < curr > nxt) or (prev > curr < nxt)):
            return False         
    return True
```

### Bài 9: Tìm và in ra tất cả các vị trí của số nguyên k trong một danh sách số nguyên cho trước. Nếu không có k trong ds thì in ra "-1".
VD: k = 3, [1,3,4,4,3,3,4,3,6,1] => k xuất hiện tại các vị trí (trả về 1 list) [1, 4, 5, 7]
k = 10, [1,3,4,4,3,3,4,3,6,1] => k không xuất hiện trong ds => trả về "-1"

```python
def tim(k, lst):
    A = [i for i, x in enumerate(lst) if x == k]
    return A if A else "-1"
k=list(input())
lst=list(input())
print(tim(k,lst))
```
