W10A2
[Invalid_Index]
Hãy nhập vào một mảng my_list (list) n số nguyên (n nguyên dương). Thực hiện thử lệnh
print(my_list[n]) rồi đọc kỹ Traceback và bắt lỗi Truy cập chỉ số sai. (Dùng try ... except
bắt lỗi IndexError)

```python
try:
    n = int(input("Nhập vào số nguyên dương n: "))
    if n <= 0:
        print("Vui lòng nhập một số nguyên dương!")
    else:
        my_list = []
        print(f"Nhập vào {n} số nguyên:")
        for i in range(n):
            element = int(input(f"Phần tử thứ {i}: "))
            my_list.append(element)
        print(f"\nThực hiện lệnh: print(my_list[{n}])")
        print(my_list[n])
except IndexError as e:
    print("\n--- BẮT ĐƯỢC LỖI ---")
    print(f"Lỗi rồi: {e}")
    print(f"Giải thích: Mảng chỉ có các chỉ số từ 0 đến {n-1}. Chỉ số {n} không tồn tại!")
except ValueError:
    print("\nLỗi: Vui lòng chỉ nhập số nguyên!")
```

W10A3
[ValueError - HappyNumbers]
Nhập vào một số nguyên dương a, Kiểm tra xem a có phải là số hạnh phúc không.
Một số nguyên dương n được gọi là số hạnh phúc nếu ta thực hiện liên tiếp các bước
sau:
- Tính tổng bình phương các chữ số của n.
- Lặp lại bước đó với kết quả vừa tính.
- Nếu cuối cùng kết quả là 1, thì n là số hạnh phúc.
- Nếu rơi vào vòng lặp vô hạn không bao giờ ra 1 (ví dụ như 4 → 16 → 37 → ...), thì n
không phải là số hạnh phúc.
Hãy nhập thử đầu vào là "a" rồi đọc kỹ Traceback để gỡ lỗi nhập liệu. (Dùng try ... except
bắt lỗi ValueError thêm set)

```python
def is_happy_number(n: int) -> bool:
    """Hàm kiểm tra xem n có phải là số hạnh phúc hay không."""
    seen = set()  
    while n != 1 and n not in seen:
        seen.add(n)
        n = sum(int(digit) ** 2 for digit in str(n))
    return n == 1
try:
    user_input = input("Nhập vào một số nguyên dương a: ")
    a = int(user_input) 
    if a <= 0:
        print("Vui lòng nhập một số nguyên DƯƠNG!")
    else:
        if is_happy_number(a):
            print(f"{a} là SỐ HẠNH PHÚC! 😊")
        else:
            print(f"{a} KHÔNG PHẢI là số hạnh phúc. 😢")
except ValueError as e:
    print("\n--- [TRACEBACK / ERROR DETECTED] ---")
    print(f"Lỗi nhập liệu: Bạn đã không nhập đúng định dạng số nguyên dương.")
    print(f"Chi tiết lỗi hệ thống: {e}")
    print("--------------------------------------")
```

W10A4
Sử dụng try except,kết hợp cùng else... finally.
Tính tổng 2 số, nếu nhập sai kiểu thì thông báo,
luôn luôn in "Kết thúc chương trình." sau khi chạy.

```python
try:
    num1 = float(input("Nhập số thứ nhất: "))
    num2 = float(input("Nhập số thứ hai: "))
except ValueError:
    print("Lỗi: Bạn phải nhập vào một số hợp lệ!")
else:
    tong = num1 + num2
    print(f"Tổng của hai số là: {tong}")
finally:
    print("Kết thúc chương trình.")
```

Viết chương trình nhập vào số tuổi của một người. Nếu số tuổi nhỏ hơn 0 hoặc không
phải số, in ra "Tuoi khong hop le". Nếu số tuổi hợp lệ thì in ra năm sinh của người ta (tính
từ năm 2025 trở về).

```python
def tinh_nam_sinh():
    input_data = input("Nhập vào số tuổi: ")   
    try:
        tuoi = int(input_data)
        if tuoi < 0:
            print("Tuoi khong hop le")
        else:
            nam_sinh = 2025 - tuoi
            print(f"Năm sinh của người đó là: {nam_sinh}")       
    except ValueError:
        print("Tuoi khong hop le")
tinh_nam_sinh()
```

W10A10
Viết chương trình nhập vào phần tử N (N > 0) và một mảng gồm N số nguyên dương.
Một mảng là hợp lệ nếu nó gồm các số khác nhau. In ra tổng các phần tử nếu mảng đó
là hợp lệ. Nếu mảng không hợp lệ, in ra "Mang khong hop le".

```python
def check_and_sum_array():
    try:
        N = int(input("Nhập số lượng phần tử N (N > 0): "))
        if N <= 0:
            print("Mang khong hop le")
            return
    except ValueError:
        print("Mang khong hop le")
        return
    try:
        arr = list(map(int, input(f"Nhập {N} số nguyên dương: ").split()))
    except ValueError:
        print("Mang khong hop le")
        return
    if len(arr) != N or any(x <= 0 for x in arr):
        print("Mang khong hop le")
        return
    if len(arr) == len(set(arr)):
        print(sum(arr))
    else:
        print("Mang khong hop le")
if __name__ == "__main__":
    check_and_sum_array()
```

W10A14
[TypeError - SumNumbers]
Nhập vào một danh sách gồm nhiều giá trị. Hãy cộng tất cả phần tử sau khi chuyển
được sang int. Nếu phần tử nào không chuyển được thì bỏ qua. In ra tổng cuối cùng.

```python
def sum_numbers(data_list):
    tong = 0
    for item in data_list:
        try:
            # Cố gắng chuyển đổi sang kiểu số nguyên và cộng vào tổng
            tong += int(item)
        except (ValueError, TypeError):
            # Nếu không chuyển được (ví dụ: chuỗi chữ, danh sách, None...), bỏ qua
            continue
    return tong
```
