Viết hàm kiểm tra một số có phải số Amstrong hay không? Hàm trả về kiểu bool, có tham số là một số
nguyên x

```python
def armstrong(x: int) -> bool:
    if x < 0:
        return False
    s = str(x)
    n = len(s)
    tong_luy_thua = sum(int(digit) ** n for digit in s)
    return tong_luy_thua == x
```

W5A20 Viết hàm detect_number_type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại 'neither'.

```python
def detect_number_type(s):
    try:
        int(s)
        return 'int'
    except ValueError:
        pass
    try:
        float(s)
        return 'float'
    except ValueError:
        pass
    return 'neither'
```

W5A21 Viết hàm is_one_product_of_others(a, b, c) trả về True nếu một trong ba số bằng tích của hai số còn lại

```python
def is_one_product_of_others(a, b, c):
    if a == b * c or b == a * c or c == a * b:
        return True
    else:
        return False
```

W5A23 Viết hàm quadratic_roots_count(a, b, c) trả về số nghiệm thực của phương trình a*x^2 + b*x + c = 0.

```python
def quadratic_roots_count(a, b, c):
    delta = b**2 - 4*a*c
    if delta > 0:
        return 2 
    elif delta == 0:
        return 1 
    else:
        return 0  
```

W5A24 Viết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không.

```python
def are_collinear(x1, y1, x2, y2, x3, y3):
    return (y2 - y1) * (x3 - x2) == (y3 - y2) * (x2 - x1)
```

Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay
không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. VD: quad = [(0,0), (4,0), (4,3),
(0,3)], p = (2,1)

```python
def point_in_quadrilateral(quad, p):
    n = len(quad)
    xp, yp = p
    sides = []
    for i in range(n):
        x1, y1 = quad[i]
        x2, y2 = quad[(i + 1) % n] 
        val = (x2 - x1) * (yp - y1) - (y2 - y1) * (xp - x1)
        sides.append(val)
    is_all_non_negative = all(s >= 0 for s in sides)
    is_all_non_positive = all(s <= 0 for s in sides)
    return is_all_non_negative or is_all_non_positive
```

W5A28 Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3).

```python
def tinh_goc_abc(x1, y1, x2, y2, x3, y3):
    ux, uy = x1 - x2, y1 - y2
    vx, vy = x3 - x2, y3 - y2
    dot_product = ux * vx + uy * vy
    a = math.sqrt(ux**2 + uy**2)
    b = math.sqrt(vx**2 + vy**2)
    if a == 0 or b == 0:
        return 0.0
    cos_theta = max(-1.0, min(1.0, dot_product / (a * b)))
    angle_rad = math.acos(cos_theta)
    angle_deg = math.degrees(angle_rad)

    return angle_deg
```

W5A30 Viết hàm tính hệ số nhị thức C(n, k).

```python
def he_so_nhi_thuc(n, k):
    if k < 0 or k > n:
        return 0
    if k == 0 or k == n:
        return 1
    if k > n // 2:
        k = n - k
    a = 1
    for i in range(k):
        a = a * (n - i) // (i + 1)
    return a
```
