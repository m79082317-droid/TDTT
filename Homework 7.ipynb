Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.

```python
def solve_beautiful_array(arr):
    counts = {}
    for x in arr:
        if x in counts:
            counts[x] += 1
        else:
            counts[x] = 1   
    y = 0
    for x in counts:
        count_x = counts[x]
        if x == 0:
            y += count_x
        elif count_x >= x:
            y += (count_x - x)
        else:
    if y == 0:
        return "Mảng đẹp!", 0
    else:
        return "Cần xóa:",y
```

Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore.
Nếu không có phần tử nào thỏa mãn thì không in gì

```python
def find_majority_element(nums):
    if not nums:
        return
    candidate = None
    count = 0
    
    for num in nums:
        if count == 0:
            candidate = num
            count = 1
        elif num == candidate:
            count += 1
        else:
            count -= 1
    if nums.count(candidate) > len(nums) // 2:
        print(candidate)
```

W7A19
Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không
dùng bất cứ thuật toán sắp xếp nào

```python
def get_element(arr, index, is_ascending):
    return arr[index] if is_ascending else arr[len(arr) - 1 - index]
def merge_sorted_arrays(arr1, arr2):
    asc1 = len(arr1) < 2 or arr1[0] <= arr1[-1]
    asc2 = len(arr2) < 2 or arr2[0] <= arr2[-1]   
    n1, n2 = len(arr1), len(arr2)
    i, j = 0, 0
    result = []
    while i < n1 and j < n2:
        val1 = get_element(arr1, i, asc1)
        val2 = get_element(arr2, j, asc2)
        if val1 <= val2:
            result.append(val1)
            i += 1
        else:
            result.append(val2)
            j += 1
    while i < n1:
        result.append(get_element(arr1, i, asc1))
        i += 1
    while j < n2:
        result.append(get_element(arr2, j, asc2))
        j += 1    
    return result
```

Cho N là ngày, T là tháng năm 2020, Tính xem đó là ngày thứ bao nhiêu
của năm đó nếu đó là ngảy hợp lệ

```python
def tinh_ngay_trong_nam(n, t):
    days_in_month = [31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    if t < 1 or t > 12:
        return "Tháng không hợp lệ (phải từ 1-12)."
    if n < 1 or n > days_in_month[t-1]:
        return f"Ngày không hợp lệ (Tháng {t} năm 2020 chỉ có {days_in_month[t-1]} ngày)."
    so_thu_tu = 0
    for i in range(t - 1): # Chạy từ tháng 0 đến tháng (t-2)
        so_thu_tu += days_in_month[i]
    so_thu_tu += n   
    return so_thu_tu
```

Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất.
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất

```python
n = int(input())
names = []
for _ in range(n):
    names.append(input().strip())
count_map = {}
for name in names:
    if name in count_map:
        count_map[name] += 1
    else:
        count_map[name] = 1
res_name = ""
max_freq = 0
for name in count_map:
    freq = count_map[name]
    if freq > max_freq:
        max_freq = freq
        res_name = name
    elif freq == max_freq:
        if res_name == "" or name < res_name:
            res_name = name
print(res_name)
```

Đếm số lượng tên trong xâu kí tự. Tên được định nghĩa là xâu có độ dài
>=2 và có chữ cái đầu viết hoa, các chữ cái khác viết thường, chỉ có kí
tự cái.

```python
def count_names(s):
    words = s.split()
    count = 0
    for word in words:
        if len(word) >= 2:
            if word[0].isupper() and word[1:].islower() and word.isalpha():
                count += 1
                
    return count
```

W7A24 Cho xâu kí tự, xóa các từ giống nhau liền kề, chỉ giữ lại từ đầu tiên

```python
def xoa_tu_lap_lien_ke(s):
    words = s.split()
    if not words:
        return ""
    ket_qua = [words[0]]
    for i in range(1, len(words)):
        if words[i] != words[i-1]:
            ket_qua.append(words[i])
    return " ".join(ket_qua)
```

W7A25
Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi
đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số

```python
def find_largest_number(number_str, k):
    n = len(number_str)
    to_remove = n - k
    stack = []
    for digit in number_str:
        while to_remove > 0 and stack and stack[-1] < digit:
            stack.pop()
            to_remove -= 1
        stack.append(digit)
    return "".join(stack[:k])
```

W7A27
Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ
hay không

```python
def check_valid_brackets(s):
    stack = []
    mapping = {")": "(", "}": "{", "]": "["}
    for char in s:
        if char in mapping:
            top_element = stack.pop() if stack else '#'
            if mapping[char] != top_element:
                return False
        else:
            stack.append(char)
```

W7A28
Bà̀i toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên
nào <= N (N là số nguyên dương cho trước). Nếu có nhiề

```python
def solve_longest_composite_sequence(n):
    if n < 4:
        return 0, []
    is_prime = [True] * (n + 1)
    is_prime[0] = is_prime[1] = False
    for p in range(2, int(n**0.5) + 1):
        if is_prime[p]:
            for i in range(p * p, n + 1, p):
                is_prime[i] = False
    primes = [i for i, prime in enumerate(is_prime) if prime]
    max_len = 0
    sequences = []
    for i in range(len(primes) - 1):
        start_num = primes[i] + 1
        end_num = primes[i+1] - 1
        current_len = end_num - start_num + 1
        
        if current_len > max_len:
            max_len = current_len
            sequences = [(start_num, end_num)]
        elif current_len == max_len and current_len > 0:
            sequences.append((start_num, end_num))
    if primes:
        last_start = primes[-1] + 1
        last_end = n
        last_len = last_end - last_start + 1
        if last_len > max_len:
            max_len = last_len
            sequences = [(last_start, last_end)]
        elif last_len == max_len and last_len > 0:
            sequences.append((last_start, last_end))
    return max_len, sequences
```

W7A30
Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập

```python
def find_lis(arr):
    n = len(arr)
    if n == 0:
        return []
    L = [1] * n
    P = [-1] * n
    for i in range(1, n):
        for j in range(i):
            if arr[i] > arr[j] and L[i] < L[j] + 1:
                L[i] = L[j] + 1
                P[i] = j
    max_length = max(L)
    best_index = L.index(max_length)
    lis = []
    while best_index != -1:
        lis.append(arr[best_index])
        best_index = P[best_index]
    return lis[::-1] 
```
